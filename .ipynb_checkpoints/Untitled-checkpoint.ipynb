```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
import geopandas as gpd  # Para cargar mapas del mundo

# Cargar un dataset geográfico del mundo (naturalearth_lowres)
world = gpd.read_file(gpd.datasets.get_path('naturalearth_lowres'))

# Función para actualizar el mapa según un parámetro (ej: población)
def plot_world_map(column_name="pop_est", cmap="viridis"):
    fig, ax = plt.subplots(figsize=(15, 10))
    world.plot(column=column_name, ax=ax, legend=True, cmap=cmap)
    ax.set_title(f"Mapamundi interactivo - {column_name}")
    plt.show()

# Widget interactivo para seleccionar columna y colormap
column_selector = widgets.Dropdown(
    options=['pop_est', 'gdp_md_est', 'continent'],
    description='Variable:'
)

cmap_selector = widgets.Dropdown(
    options=['viridis', 'plasma', 'inferno', 'magma', 'cividis'],
    description='Color:'
)

# Conectar widgets a la función
widgets.interactive(plot_world_map, column_name=column_selector, cmap=cmap_selector)
```
