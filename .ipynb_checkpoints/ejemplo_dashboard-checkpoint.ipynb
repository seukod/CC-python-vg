```python
import pandas as pd
import numpy as np
import random
import matplotlib.pyplot as plot
import ipywidgets as ipw
from IPython.display import display
import functools
```

```python
df = pd.DataFrame(pd.read_csv('permisos-de-circulacion-a-noviembre.csv',sep=";"))
print(df.columns)
```

```python
def getColumnToDim(df, columnName):
    dfAux = df.groupby([columnName]).size().reset_index(name = 'COUNT')
    resp=dfAux[columnName].to_numpy()
    resp = np.append(resp,"TODO")
    return resp
```

```python
tipos_pago = getColumnToDim(df, 'Tipo_pago')
anno_permisos = getColumnToDim(df, 'anno_permiso')
```

```python
title = ipw.HTML(value="<h1><b>LOGIN</b></h1>")
txt_userName = ipw.Text(value='user name',placeholder='enter user name',
        description='User Name:', disabled=False)
txt_pass = ipw.Password(value='password',placeholder='enter password',
        description='Password:',disabled=False)
btn_login = ipw.Button(description='Login', disabled=False,  
            button_style='success', tooltip='Login', icon='check')
login_valid = ipw.Valid(value=False)
login_valid.layout.vissible = "hidden"
login_row = ipw.HBox([btn_login,login_valid])
```

```python
login1=ipw.HBox([title,txt_userName,txt_pass,btn_login])
login1
```

```python
login2=ipw.VBox([title,txt_userName,txt_pass,btn_login])
login2
```

```python
def showInvalid(b):
    if(txt_userName.value=="lvc" and txt_pass.value=="lvc"):
        login_valid.value = True
        login_valid.layout.visibility = "visible"
    else:    
        login_valid.value = False
        login_valid.layout.visibility = "visible"

btn_login.on_click(showInvalid)

login3=ipw.VBox([title,txt_userName,txt_pass,login_row],
                layout=ipw.Layout(
                display='flex',
                flex_flow='column',
                border='solid 2px',
                align_items='stretch',
                width='30%'
            ))

login3
```

```python
sel_annos = ipw.Dropdown(options = anno_permisos,value='TODO', description='Años:')
output2 = ipw.Output(layout={'border': '1px solid black'})
cuadro1 = ipw.HTML(value="0")
```

```python
def on_change(sel_value):
    with output2:
        output2.clear_output()
        sel=sel_value['new']
        monto_pagado=0
        if(sel=="TODO"):
            monto_pagado = df["MONTO_PAGADO"].sum()
            reajuste = df["REAJUSTE"].sum()
            multa = df["MULTA"].sum()
        else:         
            dfAux = df[df['anno_permiso']==int(sel)]
            monto_pagado = dfAux["MONTO_PAGADO"].sum()
            reajuste = dfAux["REAJUSTE"].sum()
            multa = dfAux["MULTA"].sum()

        #print(monto_total)
        cuadro1.value = """<center><table border='1'>
        <tr><td><b>Monto Pagado</b></td></tr>
        <tr><td>"""+str(monto_pagado)+"""</td></tr>
        <tr><td><b>Reajuste</b></td></tr>
        <tr><td>"""+str(reajuste)+"""</td></tr>
        <tr><td><b>Multa</b></td></tr>
        <tr><td>"""+str(multa)+"""</td></tr>
        </table></center>"""
```

```python
sel_annos.observe(on_change, names='value')
```

```python
metricas = ipw.VBox( [ ipw.HBox([sel_annos]),ipw.HBox([cuadro1]) ] )
```

```python
def gpie_sexo_tipoPago(value):
    df_sex = df.groupby(['SEXO']).size()
    if(value != "TODO"):
        df_sex = df[df['Tipo_pago']==value].groupby(['SEXO']).size()
        
    df_sex.plot(kind="pie", autopct='%1.1f%%', fontsize = 8)
    plot.title('SEXO')
    plot.show()
```

```python
interact_gpie_sexo_tipoPago = ipw.interactive(gpie_sexo_tipoPago, value=tipos_pago)
```

```python
def glines_sexo_fechaVen():
    dfAux = df.groupby(['FECHA_VENCIMIENTO']).size().reset_index(name = 'COUNT')
    dfAux['FECHA_VENCIMIENTO'] = pd.to_datetime(dfAux['FECHA_VENCIMIENTO'], format="%d/%m/%Y" )
    dfAux = dfAux.sort_values(by='FECHA_VENCIMIENTO')
    eje_x = dfAux['FECHA_VENCIMIENTO'].to_numpy()
    h_y = np.zeros(len(eje_x))
    m_y = np.zeros(len(eje_x))
    
    dfAux = df.groupby(['FECHA_VENCIMIENTO',"SEXO"]).size().reset_index(name = 'COUNT')
    dfAux['FECHA_VENCIMIENTO'] = pd.to_datetime(dfAux['FECHA_VENCIMIENTO'], format="%d/%m/%Y" )
    dfAux = dfAux.sort_values(by='FECHA_VENCIMIENTO')

    for idx, x in enumerate(eje_x):
        for index, row in dfAux.iterrows():
            if(row["FECHA_VENCIMIENTO"]==x):
                if(row["SEXO"]=="H"):
                    h_y[idx]=row["COUNT"]
                else:
                    m_y[idx]=row["COUNT"]
    # plot lines
    plot.plot(eje_x, h_y, label = "H")
    plot.plot(eje_x, m_y, label = "M")
    plot.legend()
    plot.show()
```

```python
interact_glines_sexo_fechaVen = ipw.interactive(glines_sexo_fechaVen, value=tipos_pago)
```

```python
#sel_tiposPagos = ipw.Dropdown(options=tipos_pago, value='TODO', description='Number:', disabled=False,)
title_tab11 = ipw.HTML(value="<h3><b>Métricas</b></h3>") 
title_tab12 = ipw.HTML(value="<h3><b>Gráfico interactivo de (Tipo de pago/Sexo)</b></h3>") 
title_tab13 = ipw.HTML(value="<h3><b>Gráfico (fecha vencimiento/Sexo)</b></h3>") 
tab1 = ipw.HBox([ipw.VBox([title_tab11, metricas ]),
                 ipw.VBox([title_tab12, ipw.VBox(interact_gpie_sexo_tipoPago.children)]),
                 ipw.VBox([title_tab13, ipw.VBox(interact_glines_sexo_fechaVen.children)])
                ])
tab2 = ipw.HBox([title])
```

```python
tab = ipw.Tab()
tab.children = [login3,tab1,tab2]
tab.titles = ["Login","Dimensión SEXO","Graficos 2"]
tab
```
